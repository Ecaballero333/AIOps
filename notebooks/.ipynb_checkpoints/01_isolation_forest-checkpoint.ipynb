# Deteccion de Anomalias — Isolation Forest

**Obligatorio AIOps — Equipo PharmaGo**

Este notebook implementa el algoritmo **Isolation Forest** para la deteccion no supervisada de anomalias en datos operacionales.

Isolation Forest aísla observaciones construyendo arboles de decision aleatorios. Las anomalías,
al ser puntos atípicos, requieren menos particiones para ser aisladas y por lo tanto tienen
caminos mas cortos en el arbol (puntaje de anomalia mas bajo).

---
**Contenido:**
1. Carga de datos
2. Exploracion (EDA)
3. Preprocesamiento
4. Entrenamiento — Isolation Forest
5. Resultados y metricas
6. Conclusiones

## 0. Setup

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
import time
import warnings

warnings.filterwarnings('ignore')
np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.dpi'] = 100

print('Librerías cargadas correctamente.')
````

## 1. Carga de datos

> **NOTA:** Actualmente se generan datos sintéticos que simulan métricas operacionales de PharmaGo.
> Cuando el dataset provisto por los docentes esté disponible, reemplazar la llamada a
> `generate_synthetic_data()` por:
> ```python
> df = pd.read_csv("../data/dataset_docentes.csv", parse_dates=['timestamp'])
> ```
> y eliminar la columna `ground_truth` del resto del notebook si no existe en el dataset real.

````python
def generate_synthetic_data(n_points=2016, anomaly_fraction=0.05, seed=42):
    """
    Genera datos sinteticos de metricas operacionales de PharmaGo.
    Simula 1 semana de datos con intervalos de 5 minutos (7 dias x 288 puntos/dia = 2016 puntos).

    Features:
        cpu_usage_percent    : uso de CPU del pod principal (%).
        memory_usage_percent : uso de memoria del pod principal (%).
        response_time_ms     : latencia promedio de respuesta HTTP (ms).
        http_error_rate      : tasa de errores HTTP 5xx (%).
        request_count        : cantidad de requests en el intervalo.

    Las anomalias simulan: picos de CPU/memoria, degradacion de latencia,
    tormentas de errores, y caidas de trafico.
    """
    np.random.seed(seed)
    n_anomalies = int(n_points * anomaly_fraction)

    timestamps = pd.date_range(start='2024-01-01', periods=n_points, freq='5min')

    # Metricas en condicion normal
    cpu      = np.random.normal(40, 8, n_points).clip(5, 75)
    memory   = np.random.normal(60, 10, n_points).clip(20, 80)
    rt       = np.random.normal(200, 40, n_points).clip(50, 500)
    err_rate = np.random.exponential(0.5, n_points).clip(0, 4)
    req_cnt  = np.random.normal(100, 20, n_points).clip(10, 250)

    # Inyectar anomalias
    anomaly_idx = np.random.choice(n_points, n_anomalies, replace=False)
    cpu[anomaly_idx]      = np.random.uniform(88, 100, n_anomalies)
    memory[anomaly_idx]   = np.random.uniform(90, 99, n_anomalies)
    rt[anomaly_idx]       = np.random.uniform(2000, 6000, n_anomalies)
    err_rate[anomaly_idx] = np.random.uniform(18, 45, n_anomalies)
    req_cnt[anomaly_idx]  = np.random.uniform(0, 5, n_anomalies)

    df = pd.DataFrame({
        'timestamp'            : timestamps,
        'cpu_usage_percent'    : cpu.round(2),
        'memory_usage_percent' : memory.round(2),
        'response_time_ms'     : rt.round(2),
        'http_error_rate'      : err_rate.round(4),
        'request_count'        : req_cnt.round(0).astype(int),
        'ground_truth'         : np.where(np.isin(np.arange(n_points), anomaly_idx), -1, 1)
    })
    return df


# --- PUNTO DE ENTRADA: reemplazar esta linea con pd.read_csv() cuando llegue el dataset real ---
df = generate_synthetic_data()

print(f'Shape: {df.shape}')
print(f'Rango temporal: {df.timestamp.min()}  →  {df.timestamp.max()}')
print(f'Anomalias reales (ground truth): {(df.ground_truth == -1).sum()} ({(df.ground_truth == -1).mean()*100:.1f}%)')
df.head()
````

## 2. Exploracion de datos (EDA)

````python
# Estadisticas descriptivas (excluye timestamp y ground_truth)
features = ['cpu_usage_percent', 'memory_usage_percent', 'response_time_ms',
            'http_error_rate', 'request_count']

print('=== Estadisticas descriptivas ===')
df[features].describe().round(2)
````

````python
print('=== Valores nulos ===')
print(df[features].isnull().sum())
print(f'\nDuplicados: {df.duplicated().sum()}')
````

````python
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
axes = axes.flatten()

colors = ['#4C72B0', '#DD8452', '#55A868', '#C44E52', '#8172B3']
labels = ['CPU (%)', 'Memoria (%)', 'Latencia (ms)', 'Tasa de errores (%)', 'Requests']

for i, (feat, color, label) in enumerate(zip(features, colors, labels)):
    axes[i].hist(df[feat], bins=50, color=color, alpha=0.8, edgecolor='white')
    axes[i].set_title(label, fontsize=11, fontweight='bold')
    axes[i].set_xlabel('Valor')
    axes[i].set_ylabel('Frecuencia')

axes[5].axis('off')
fig.suptitle('Distribuciones de metricas operacionales', fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
````

````python
# Series temporales de todas las metricas
fig, axes = plt.subplots(5, 1, figsize=(16, 14), sharex=True)

for i, (feat, color, label) in enumerate(zip(features, colors, labels)):
    axes[i].plot(df['timestamp'], df[feat], color=color, linewidth=0.6, alpha=0.8)
    axes[i].set_ylabel(label, fontsize=9)
    axes[i].xaxis.set_major_formatter(mdates.DateFormatter('%d/%m %H:%M'))

plt.xticks(rotation=30, fontsize=8)
fig.suptitle('Series temporales — 1 semana de metricas PharmaGo', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# Heatmap de correlaciones
plt.figure(figsize=(8, 6))
corr = df[features].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=0.5,
            xticklabels=labels, yticklabels=labels)
plt.title('Correlacion entre metricas operacionales', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 3. Preprocesamiento

Isolation Forest no es sensible a la escala, pero normalizar los datos mejora la comparabilidad
de los puntajes de anomalia entre features y facilita las visualizaciones con PCA.

````python
X = df[features].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f'Shape de la matriz de features: {X_scaled.shape}')
print(f'Media post-escalado (debe ser ~0): {X_scaled.mean(axis=0).round(4)}')
print(f'Std post-escalado (debe ser ~1):  {X_scaled.std(axis=0).round(4)}')
````

## 4. Entrenamiento — Isolation Forest

**Hiperparámetros clave:**
- `n_estimators=200`: numero de arboles. A mayor valor, mas estable el puntaje (default=100).
- `contamination=0.05`: fraccion esperada de anomalias en el dataset. Se ajusta al 5% conocido
  del dataset sintetico; con datos reales se debe estimar (tipico: 0.01–0.10 en produccion).
- `max_samples='auto'`: usa `min(256, n_samples)` puntos por arbol.
- `random_state=42`: garantiza reproducibilidad.

````python
CONTAMINATION = 0.05  # ajustar segun el dataset real

t0 = time.time()

modelo_if = IsolationForest(
    n_estimators=200,
    contamination=CONTAMINATION,
    max_samples='auto',
    random_state=42,
    n_jobs=-1
)
modelo_if.fit(X_scaled)

t_train = time.time() - t0
print(f'Entrenamiento completado en {t_train:.3f}s')

# Predicciones: -1 = anomalia, 1 = normal
predicciones_if = modelo_if.predict(X_scaled)

# Puntaje de anomalia: mas negativo = mas anomalo
scores_if = modelo_if.decision_function(X_scaled)

n_anomalias = (predicciones_if == -1).sum()
print(f'Anomalias detectadas: {n_anomalias} ({n_anomalias/len(df)*100:.2f}%)')
````

## 5. Resultados y metricas

````python
# Agregar predicciones al dataframe
df['prediccion_if'] = predicciones_if
df['score_if']      = scores_if

# Resumen de resultados
print('=== Resumen de deteccion ===')
print(f'Total de puntos analizados : {len(df)}')
print(f'Normales detectados        : {(predicciones_if == 1).sum()}')
print(f'Anomalias detectadas       : {(predicciones_if == -1).sum()}')
print(f'Tasa de anomalias          : {(predicciones_if == -1).mean()*100:.2f}%')
print(f'Tiempo de entrenamiento    : {t_train:.3f}s')
````

````python
# Metricas supervisadas (disponibles porque tenemos ground truth sintetico)
# Con el dataset real de los docentes, esta seccion puede no estar disponible si no hay etiquetas.
print('=== Metricas de clasificacion (vs. ground truth) ===')
print(classification_report(df['ground_truth'], predicciones_if,
                             target_names=['Anomalia (-1)', 'Normal (1)']))

# ROC-AUC usando scores como probabilidad
# Invertimos el signo porque scores mas negativos = mas anomalo
auc = roc_auc_score(df['ground_truth'].map({1: 0, -1: 1}),
                    -scores_if)
print(f'ROC-AUC Score: {auc:.4f}')

print('\n=== Matriz de confusion ===')
cm = confusion_matrix(df['ground_truth'], predicciones_if)
print(pd.DataFrame(cm,
                   index=['Real: Anomalia', 'Real: Normal'],
                   columns=['Pred: Anomalia', 'Pred: Normal']))
````

````python
# Histograma de puntajes de anomalia
plt.figure(figsize=(10, 4))
plt.hist(scores_if[predicciones_if == 1],  bins=60, alpha=0.7, color='steelblue', label='Normal')
plt.hist(scores_if[predicciones_if == -1], bins=30, alpha=0.8, color='crimson',   label='Anomalia')
plt.axvline(x=0, color='black', linestyle='--', linewidth=1, label='Umbral (score=0)')
plt.xlabel('Anomaly Score (decision_function)', fontsize=11)
plt.ylabel('Frecuencia', fontsize=11)
plt.title('Distribucion de puntajes de anomalia — Isolation Forest', fontsize=12, fontweight='bold')
plt.legend()
plt.tight_layout()
plt.show()
````

````python
# Series temporales con anomalias marcadas
fig, axes = plt.subplots(3, 1, figsize=(16, 10), sharex=True)

metricas_viz = [
    ('cpu_usage_percent',  'CPU (%)',         '#4C72B0'),
    ('response_time_ms',   'Latencia (ms)',   '#DD8452'),
    ('http_error_rate',    'Error rate (%)',  '#C44E52'),
]

for ax, (feat, label, color) in zip(axes, metricas_viz):
    ax.plot(df['timestamp'], df[feat], color=color, linewidth=0.7, alpha=0.8, label='Normal')

    idx_anom = df[df['prediccion_if'] == -1].index
    ax.scatter(df.loc[idx_anom, 'timestamp'], df.loc[idx_anom, feat],
               color='red', s=20, zorder=5, label='Anomalia detectada')

    ax.set_ylabel(label, fontsize=9)
    ax.legend(fontsize=8, loc='upper right')
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%d/%m'))

plt.xticks(rotation=15, fontsize=8)
fig.suptitle('Anomalias detectadas por Isolation Forest en series temporales',
             fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# Visualizacion 2D con PCA
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(9, 6))
mask_normal = predicciones_if == 1
mask_anom   = predicciones_if == -1

plt.scatter(X_pca[mask_normal, 0], X_pca[mask_normal, 1],
            c='steelblue', s=8, alpha=0.4, label=f'Normal ({mask_normal.sum()})')
plt.scatter(X_pca[mask_anom, 0], X_pca[mask_anom, 1],
            c='red', s=40, alpha=0.8, label=f'Anomalia ({mask_anom.sum()})', marker='x')

var_exp = pca.explained_variance_ratio_ * 100
plt.xlabel(f'PC1 ({var_exp[0]:.1f}% varianza)', fontsize=10)
plt.ylabel(f'PC2 ({var_exp[1]:.1f}% varianza)', fontsize=10)
plt.title('Proyeccion PCA — Isolation Forest', fontsize=12, fontweight='bold')
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()

print(f'Varianza explicada por PC1+PC2: {sum(var_exp):.1f}%')
````

````python
# Boxplots: distribucion de cada metrica segun prediccion
fig, axes = plt.subplots(1, 5, figsize=(18, 5))

for ax, feat, label in zip(axes, features, labels):
    data_normal = df.loc[df['prediccion_if'] == 1,  feat]
    data_anom   = df.loc[df['prediccion_if'] == -1, feat]
    ax.boxplot([data_normal, data_anom], labels=['Normal', 'Anomalia'],
               patch_artist=True,
               boxprops=dict(facecolor='steelblue', alpha=0.6),
               medianprops=dict(color='black', linewidth=2))
    ax.set_title(label, fontsize=9, fontweight='bold')
    ax.tick_params(axis='x', labelsize=8)

fig.suptitle('Distribucion de metricas: Normal vs. Anomalia (Isolation Forest)',
             fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 6. Conclusiones

En esta celda se documentan los hallazgos del experimento.
**Actualizar con conclusiones reales cuando se use el dataset de los docentes.**

````python
from sklearn.metrics import precision_score, recall_score, f1_score

y_true = (df['ground_truth'] == -1).astype(int)
y_pred = (df['prediccion_if'] == -1).astype(int)

print('=== Resumen final — Isolation Forest ===')
print(f'  Dataset               : {len(df)} observaciones')
print(f'  Features              : {len(features)}')
print(f'  Contaminacion usada   : {CONTAMINATION}')
print(f'  Anomalias detectadas  : {y_pred.sum()} ({y_pred.mean()*100:.2f}%)')
print(f'  Precision             : {precision_score(y_true, y_pred):.4f}')
print(f'  Recall                : {recall_score(y_true, y_pred):.4f}')
print(f'  F1-Score              : {f1_score(y_true, y_pred):.4f}')
print(f'  ROC-AUC               : {auc:.4f}')
print(f'  Tiempo de entreno     : {t_train:.3f}s')
print()
print('Observaciones:')
print('  - Los puntos anomalos presentan CPU >88%, latencia >2000ms y error rate >18% simultaneamente.')
print('  - La proyeccion PCA muestra separacion clara entre normales y anomalias en el espacio reducido.')
print('  - Isolation Forest es eficiente: entrenamiento en menos de 1 segundo para 2016 puntos.')
print('  - Con el dataset real, ajustar el parametro contamination segun la tasa historica de incidentes.')
````
